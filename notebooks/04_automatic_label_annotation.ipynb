```python
import pandas as pd
top_words_per_topic = []
for t in range(model.num_topics):
    top_words_per_topic.extend([(t, ) + x for x in model.show_topic(t, topn = 10)])

df = pd.DataFrame(top_words_per_topic, columns=['Topic', 'Word','P'])
print(df)
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
df_prod_review = pd.read_csv('https://raw.githubusercontent.com/aakashgoel12/blogs/master/input/product_review_sentiment.csv',\
                      encoding='latin-1')

all_intents = df_prod_review.Review.tolist()
sentiments = df_prod_review.user_sentiment.tolist()

all_sents = []

for intent in all_intents:
    for sent in nltk.sent_tokenize(intent):
        if len(sent.split()) > 4:
            all_sents.append(sent)

print(len(all_sents))

all_intent = all_sents
```

```python
doc_topics = [model.get_document_topics(bow) for bow in corpus_tfidf]

lda_labels = [max(doc, key=lambda x: x[1])[0] if doc else -1 for doc in doc_topics]

doc_topics
```

```python
data_clustered = pd.DataFrame(data = list(zip(all_intents, sentiments, lda_labels)) , columns=['text', 'user_sentiment', 'label_st1' ])

data_clustered.head(10)
```

```python
data_clustered = pd.DataFrame(data = list(zip(all_intents, lda_labels)) , columns=['text', 'label_st1' ] )
data_clustered['sentiment'] = df_prod_review['user_sentiment'].values

data_clustered.head()
```

```python
example_category = data_clustered[data_clustered['label_st1'] == 3].sample(frac=1, random_state=42).reset_index(drop=True)
example_category.to_csv("topik3.csv", index=False)
```
